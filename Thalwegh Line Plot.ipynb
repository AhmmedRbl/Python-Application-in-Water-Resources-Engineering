```python
import pandas as pd
import os

folder_path = r"F:\P-529990 Joydeppur Iswardi Double Rail Line\CS Section Data Khal"

data_df = {}   # dictionary to store all DataFrames

for file in os.listdir(folder_path):
    if file.lower().endswith(".xlsm"):
        base_name = os.path.splitext(file)[0]   # automatic key
        file_path = os.path.join(folder_path, file)

        data_df[base_name] = pd.read_excel(
            file_path,
            engine="openpyxl"
        )

print(data_df.keys())
```

```python
for name, df in data_df.items():
    print(name)
    display(df.head())
```

```python
import os

output_folder = r"F:\P 529990 Joydepur Iswardi Double Rail Line\CS_Min_Elevation_Output"
os.makedirs(output_folder, exist_ok=True)

for name, df in data_df.items():
    
    # group by Chainage and take minimum Elevation
    min_elev_df = (
        df.groupby("Chainage", as_index=False)
          .agg({"Elevation": "min"})
    )
    
    # save to Excel (one file per DataFrame)
    output_path = os.path.join(output_folder, f"{name}_min_elevation.xlsx")
    min_elev_df.to_excel(output_path, index=False)

print("✅ Min elevation Excel created for all CS sections")
```
