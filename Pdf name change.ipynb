```python
import os
from PyPDF2 import PdfReader, PdfWriter
import re

# Folder where your PDF files are stored
input_folder = r"F:\P-52797\PDF Name Change\Main Report"   # 🔹 change this path

# Output folder to save new PDFs
output_folder = os.path.join(input_folder, "Top_Page_Reports")
os.makedirs(output_folder, exist_ok=True)

for filename in os.listdir(input_folder):
    if filename.lower().endswith(".pdf") and filename.startswith("Main Report_PS-"):
        input_path = os.path.join(input_folder, filename)

        try:
            # Read the original PDF
            reader = PdfReader(input_path)
            writer = PdfWriter()

            # Add only the first page
            if len(reader.pages) > 0:
                writer.add_page(reader.pages[0])

                # Extract the number after PS-
                match = re.search(r'PS-(\d+)', filename)
                if match:
                    num_part = match.group(1)
                else:
                    num_part = "Unknown"

                # New filename
                new_filename = f"Top Page_Final Report_PS-{num_part}.pdf"

                output_path = os.path.join(output_folder, new_filename)

                # Write the new single-page PDF
                with open(output_path, "wb") as out_file:
                    writer.write(out_file)

                print(f"✅ Saved: {new_filename}")
            else:
                print(f"⚠️ Skipped empty file: {filename}")

        except Exception as e:
            print(f"❌ Error with {filename}: {e}")

print("\n🎉 All first pages saved successfully!")
```
